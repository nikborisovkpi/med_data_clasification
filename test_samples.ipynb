```python
from pathlib import Path
import numpy as np
from PIL import Image

# Исходные пути
data_dir = Path("data/medmnist/organamnist_224")
images = np.load(data_dir / "test_images.npy", mmap_mode="r")
labels = np.load(data_dir / "test_labels.npy").reshape(-1)

# Маппинг оригинальных меток OrganAMNIST:
# 6: liver, 4: kidney-left, 7: lung-left, 3: heart (other)
targets = {
    "liver": 6,
    "kidney": 4,
    "lung": 7,
    "other_heart": 3,
}

out_dir = Path("test_samples")
out_dir.mkdir(exist_ok=True)

for name, raw_label in targets.items():
    idx = int(np.flatnonzero(labels == raw_label)[0])
    img_array = images[idx]
    if img_array.ndim == 3 and img_array.shape[-1] == 1:
        img_array = img_array[..., 0]
    
    img = Image.fromarray(img_array.astype(np.uint8), mode="L")
    img.save(out_dir / f"sample_{name}.png")
    print(f"Сохранено: test_samples/sample_{name}.png (индекс {idx})")
```
